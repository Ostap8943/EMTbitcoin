```python
import pandas as pd

import datetime
import plotly.graph_objects as go
```

```python
btc = pd.read_csv("../Dataset/btcusd_1-min_data.csv")
```

```python
btc['datetime'] = pd.to_datetime(btc['Timestamp'],unit='s')
btc['date'] = pd.to_datetime(btc['datetime']).dt.date
btc['time'] = pd.to_datetime(btc['datetime']).dt.time


firstday = btc[btc['date'] == datetime.date(2021, 2, 10)]
```

```python
firstday
```

```python
fig = go.Figure(data=go.Ohlc(x=firstday['time'],
                    open=firstday['Open'],
                    high=firstday['High'],
                    low=firstday['Low'],
                    close=firstday['Close']))
fig.show()
```
