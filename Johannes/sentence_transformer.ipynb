```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
import torch
```

```python
df_sentenc_input = pd.read_parquet("bitcoin_users.parquet", columns=['user_name', 'text', 'user_description'])
```

```python
df_sentenc_input['input_text'] = (df_sentenc_input['user_description'].fillna('') + ' ' + df_sentenc_input['text'].fillna('')).str.strip()
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

model = SentenceTransformer("all-mpnet-base-v2", device=device)
print(model.device)
```

```python
embeddings = np.stack(df_sentenc_input["input_text"].apply(model.encode).values)
```

```python
np.savez_compressed("user_embeddings.npz", embeddings=embeddings, user_names=df_sentenc_input["user_name"].values)
```
